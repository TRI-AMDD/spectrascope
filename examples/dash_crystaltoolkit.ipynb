```python
import dash 
from dash import html, dcc
import crystal_toolkit

import crystal_toolkit.components as ctc
from pymatgen.core.structure import Structure
import pandas as pd 
import numpy as np 

top10 = ['mp-603612', 'mp-6539', 'mp-6706', 'mp-554111', 'mp-6125',
       'mp-554704', 'mp-560586', 'mp-984604', 'mp-6268', 'mp-561121']
structures_df = pd.read_json(('/Users/amalyajohnson/Desktop/Spectrascope//Datasets/XAS_PDF/Billinge_dataset/Ti_ss_class_structures_df.json'))
top_structures = [Structure.from_dict(structures_df.loc[mpid]['structure']) for mpid in top10]
app = dash.Dash()


structure_components = [ctc.StructureMoleculeComponent(struct, id=f"structure_{i}") for i, struct in enumerate(top_structures)]

test = ctc.StructureMoleculeComponent(top_structures[0], id='hi')


my_layout = html.Div(className='row', children=[
    html.H1("Top 10 "),
    dcc.Dropdown(),
    html.Div(children=[
        struct.layout() for struct in structure_components
    ])
])

ctc.register_crystal_toolkit(app=app, layout=my_layout)

if __name__ == "__main__":
    app.run_server(debug=True, port=8080)
```
